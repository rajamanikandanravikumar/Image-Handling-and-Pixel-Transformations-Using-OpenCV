10. Read in the image ('Apollo-11-launch.jpg').

```python
apollo = cv2.imread('Apollo-11-launch.jpg')
if apollo is None:
    raise FileNotFoundError("Apollo-11-launch.jpg not found")
apollo = cv2.cvtColor(apollo, cv2.COLOR_BGR2RGB)
plt.imshow(apollo)
plt.title("Apollo 11 Launch")
plt.axis('off')
plt.show()
```

11. Add the following text to the dark area at the bottom of the image (centered on the image):

```python
text = 'Apollo 11 Saturn V Launch, July 16, 1969'
font_face = cv2.FONT_HERSHEY_PLAIN
font_scale = 1.5
thickness = 2
apollo_bgr = cv2.cvtColor(apollo.copy(), cv2.COLOR_RGB2BGR)
text_size, baseline = cv2.getTextSize(text,font_face,font_scale,thickness)
text_x = (apollo.shape[1] - text_size[0]) // 2
text_y = apollo.shape[0] - 20
cv2.putText(apollo_bgr,text,(text_x, text_y),font_face,font_scale,(255, 255, 255),thickness,cv2.LINE_AA)
apollo_annotated = cv2.cvtColor(apollo_bgr, cv2.COLOR_BGR2RGB)
```

12. Draw a magenta rectangle that encompasses the launch tower and the rocket.

```python
# rect_color = magenta
# YOUR CODE HERE
rect_color = (255, 0, 255)  
x1 = int(apollo.shape[1] * 0.35)
y1 = int(apollo.shape[0] * 0.10)
x2 = int(apollo.shape[1] * 0.70)
y2 = int(apollo.shape[0] * 0.90)
cv2.rectangle(apollo_annotated,(x1, y1),(x2, y2),rect_color,3)
```

13. Display the final annotated image.

```python
# 13. Display the final annotated image
plt.figure(figsize=(12, 8))
plt.imshow(apollo_annotated)
plt.title("Apollo 11 - Annotated Image")
plt.axis('off')
plt.show()
cv2.imwrite(
    'Apollo_11_Annotated.png',
    cv2.cvtColor(apollo_annotated, cv2.COLOR_RGB2BGR)
)
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
boy_bgr = cv2.cvtColor(boy,cv2.COLOR_RGB2BGR)
B, G, R = cv2.split(boy_bgr)
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(B, cmap='gray')
plt.title("Blue Channel")
plt.axis('off')
plt.subplot(1, 3, 2)
plt.imshow(G, cmap='gray')
plt.title("Green Channel")
plt.axis('off')
plt.subplot(1, 3, 3)
plt.imshow(R, cmap='gray')
plt.title("Red Channel")
plt.axis('off')
plt.tight_layout()
plt.show()
```

21. _merged the R, G, B , display along with orginal image 

```python
merged_rgb = cv2.merge((R, G, B))

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(boy)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged_rgb)
plt.title("Merged RGB Image")
plt.axis('off')

plt.tight_layout()
```

22. Split the image into the H, S, V components &  Display the channels.

```python
hsv = cv2.cvtColor(
    boy,
    cv2.COLOR_RGB2HSV
)

H, S, V = cv2.split(hsv)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(H, cmap='gray')
plt.title("Hue Channel")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(S, cmap='gray')
plt.title("Saturation Channel")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(V, cmap='gray')
plt.title("Value Channel")
plt.axis('off')

plt.tight_layout()

plt.show()
```

23. _merged the H, S, V, display along with orginal image 

```python
merged_hsv = cv2.merge((H, S, V))

merged_hsv_rgb = cv2.cvtColor(
    merged_hsv,
    cv2.COLOR_HSV2RGB
)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(boy)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(merged_hsv_rgb)
plt.title("Merged HSV Image")
plt.axis('off')

plt.tight_layout()

plt.show()
```
